```python
import pandas as pd
df=pd.read_csv(r"C:\Users\ABINAYA ANBAZHAGAN\Downloads\archive\student_lifestyle_dataset.csv")
df.head()
```

```python
# Count missing values in each column
print(df.isnull().sum())

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Display updated DataFrame
print(df)
```

```python
df.columns
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(df['Study_Hours_Per_Day'], bins=10)

plt.title("Distribution of Study Hours")
plt.xlabel("Study Hours Per Day")
plt.ylabel("Number of Students")

plt.show()
# Most students study between moderate hours per day.
```

```python
sns.scatterplot(
    x='Study_Hours_Per_Day',
    y='GPA',
    data=df
)

plt.title("Study Hours vs GPA")

plt.show()
# Students who study more hours tend to have a higher GPA.
```

```python
sns.boxplot(y='Sleep_Hours_Per_Day', data=df)

plt.title("Sleep Hours Distribution")

plt.show()
# Most students maintain an average sleep schedule.
```

```python
sns.scatterplot(
    x='Social_Hours_Per_Day',
    y='GPA',
    data=df
)

plt.title("Social Hours vs GPA")

plt.show()
# Excessive social hours may slightly reduce GPA.
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
# The heatmap shows relationships between lifestyle habits and GPA.
```

```python
sns.countplot(x='Stress_Level', data=df)

plt.title("Stress Level Distribution")

plt.show()
# Many students experience moderate to high stress levels.
```

```python
sns.barplot(
    x='Stress_Level',
    y='GPA',
    data=df
)

plt.title("Average GPA by Stress Level")

plt.show()
# Students with lower stress levels tend to perform better academically
```

```python
## Conclusion

- Study hours positively impact GPA.
- Balanced sleep improves performance.
- High stress levels may reduce academic performance.
- Social activities should be balanced with study time.
```
